# ARRAY MAUPLATION FUNCTIONS 

```python
a1 = np.array([1,2,3])
reshaped = np.reshape(a1 ,(1,3))    # reshape 1 x 3
print("Reshaped array:",reshaped)
```

```python
f1 = np.array([[1,2],[3,4]])
flattened = np.ravel(f1)     # flattened to 1d array 
print('flattend array:',flattened)
```

```python
e1 = np.array([[1,2] , [3,4]])
transposed = np.transpose(e1)        # transpose the array 
print("transposed array:\n",transposed)
```

```python
a2 = np.array([1,2])
b2 = np.array([3,4])
stacked = np.vstack([a2,b2])    # stack a nad b are vertically 
print("stacked array:\n",stacked)
```

# STATISTICAL FUNCTIONS 

```python
s = np.array([1,2,3,4])  # average 
mean = np.mean(s)
print("Mean of s:",mean)
```

```python
std_dev = np.std(s)         # standard deviation 1 - 2.5 square dived by element number 
print("Standard deviation of s:", std_dev)
```

```python
# min eleemnt of an array 
minimum = np.min(s)
print("standard deviation of s:",std_dev)
```

```python
minimum = np.min(s)
print("min of s:",minimum)
```

```python
maximum = np.max(s)
print("max of s:",maximum)
```

# LINEAR ALGEBRA FUNCTIONS 

```python
# CREATE  a matrix
matrix = np.array([[1,2],[3,4]])
```

```python
# determinant of matrix 
determinant = np.linalg.det(matrix)
print("Determinant of matrix:",determinant)
```

```python
# Inversse of a matrix 
inverse = np.linalg.inv(matrix)
print("Inverse of matrix:\n" , inverse)
```

# BOOLEAN & LOGICAL FUNCTIONS 

```python
# check if all elements are true 
# all
logical_test = np.array([True,False,True])
all_true = np.all(logical_test)
print("All elements are True:",all_true)
```

```python
logical_test = np.array([False,False,False])
all_true = np.all(logical_test)
print("All elements are True:",all_true)
```

```python
any_true = np.any(logical_test)
print("Any elements are True:",any_true)
```

# SET OPERATIONS 

```python
# INTERSECTION of two arrays 
set_a = np.array([1,2,3,4])
set_b = np.array([3,4,5,6])
intersection = np.intersect1d(set_a,set_b)
print("intersection of a and b:", intersection)
```

```python
union = np.union1d(set_a,set_b)
print("Union of a and b:",union)
```

# ARRAY ATTRIBUTE FUNCTIONS

```python
# array attributes
a = np.array([1,2,3])
shape = a.shape    # shape of the array 
size = a.size        # numver of elements 
dimensions = a.ndim    # number of dimensions 
dtype = a.dtype  # datatype of the array 
print("shape of a:",shape)
print("size of a:",size)
print("Number of dimensionsof a:",dimensions)
print("Data type of a:",dtype)
```
